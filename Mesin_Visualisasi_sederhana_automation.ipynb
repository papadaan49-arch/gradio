**Reasoning**:
Since I cannot directly access the user's specific sheet, I will create a dummy DataFrame to allow the workflow to continue. This DataFrame will serve as a placeholder that the user can replace with their actual data loading code.



````python
import pandas as pd
import numpy as np

# Create a dummy DataFrame as a placeholder
data = {
    'Category': ['A', 'B', 'A', 'C', 'B', 'C', 'A', 'B'],
    'Value': np.random.randint(10, 100, 8),
    'Date': pd.to_datetime(['2023-01-01', '2023-01-02', '2023-01-03', '2023-01-04', '2023-01-05', '2023-01-06', '2023-01-07', '2023-01-08'])
}
df = pd.DataFrame(data)

print("Dummy DataFrame created successfully. Please replace this with your actual data loading code.")
print("First 5 rows of the DataFrame:")
print(df.head().to_markdown(index=False))
````

## Generate Visualization

### Subtask:
Membuat contoh visualisasi menggunakan data yang dimuat. Anda bisa memodifikasi kode ini untuk membuat berbagai jenis plot (misalnya, bar chart, line plot, scatter plot) sesuai dengan kebutuhan analisis Anda. Pastikan untuk memberikan legenda pada visualisasi.


**Reasoning**:
I will import the necessary visualization libraries, `matplotlib.pyplot` and `seaborn`, to prepare for creating the bar plot as requested in the instructions.



````python
import matplotlib.pyplot as plt
import seaborn as sns

print("Matplotlib and Seaborn libraries imported successfully.")
````

**Reasoning**:
I will create a bar plot using seaborn to visualize the average 'Value' for each 'Category' from the DataFrame, add appropriate labels and a title, and display the plot.



````python
plt.figure(figsize=(8, 6))
sns.barplot(x='Category', y='Value', data=df.groupby('Category')['Value'].mean().reset_index())
plt.xlabel('Category')
plt.ylabel('Average Value')
plt.title('Average Value per Category')
plt.show()
````

## Download as PNG/JPG

### Subtask:
Menyediakan kode untuk menyimpan visualisasi yang telah dibuat ke dalam format gambar PNG dan JPG, sehingga Anda bisa mengunduhnya.


**Reasoning**:
I will save the last generated visualization as a PNG file and then as a JPG file with a specified DPI. After saving, I will call `plt.show()` to display the plot in the output, as requested in the instructions.



````python
import matplotlib.pyplot as plt

# Save the visualization as PNG
plt.savefig('visualisasi.png')
print("Visualization saved as visualisasi.png")

# Save the visualization as JPG with specified DPI
plt.savefig('visualisasi.jpg', dpi=300)
print("Visualization saved as visualisasi.jpg")

plt.show()
````
